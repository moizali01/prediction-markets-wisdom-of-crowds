# Polymarket EDA

## Preamble

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json
import sqlite3
```

## Load data

```python
# Open the database
con = sqlite3.connect("../polymarket-scraper/events.db")
```

```python
# Read SQL db into DataFrame
df = pd.read_sql("SELECT * FROM events",con)
```

## Clean Data

```python
df.info()
```

```python
# Set the correct type for boolean columns
boolean_columns = (
    "active",
    "closed",
    "archived",
    "new",
    "featured",
    "restricted",
    "commentsEnabled",
    "enableOrderBook",
    "negRisk",
    "cyom",
    "showAllOutcomes",
    "showMarketImages",
    "automaticallyResolved",
    "enableNegRisk",
    "automaticallyActive",
    "live",
    "ended",
    "estimateValue",
    "cantEstimate",
    "pendingDeployment",
    "deploying"
)

df[[*boolean_columns]] = df[[*boolean_columns]].astype('boolean')
```

```python
# Parse the string JSON values
object_columns = (
    "markets",
    "tags",
    "eventCreators",
)
```

```python
def parse_json(json_str : str) : 
    return json.loads(json_str) if isinstance(json_str,str) else np.nan
    
df[[*object_columns]] = df[[*object_columns]].map(parse_json)
```
